```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
import os

# 1. LOAD THE CDC NHANES 2013-2014 DATASETS
print("Locating data files...")
# Automatically detects where Colab placed your uploads
path_prefix = '/' if os.path.exists('/DEMO_H.xpt') else '/content/'

df_demo = pd.read_sas(f'{path_prefix}DEMO_H.xpt')
df_ghb  = pd.read_sas(f'{path_prefix}GHB_H.xpt')
df_smq  = pd.read_sas(f'{path_prefix}SMQ_H.xpt')
df_ohx  = pd.read_sas(f'{path_prefix}OHXPER_H.xpt')

print("Applying strict epidemiological filters...")

# 2. ISOLATE DEMOGRAPHIC AND LABORATORY VARIABLES
df_demo = df_demo[['SEQN', 'RIDAGEYR', 'RIAGENDR']]
df_ghb = df_ghb[['SEQN', 'LBXGH']]

# 3. SMOKING STATUS: CURRENT vs NEVER (former smokers excluded)
condition_current = (df_smq['SMQ020'] == 1) & (df_smq['SMQ040'].isin([1, 2]))
condition_never = (df_smq['SMQ020'] == 2)

df_smq_filtered = df_smq[condition_current | condition_never].copy()

# Apply the 1 or 0 binary label
df_smq_filtered['current_smoker'] = np.where(
    (df_smq_filtered['SMQ020'] == 1) & (df_smq_filtered['SMQ040'].isin([1, 2])),
    1,
    0
)
df_smq_clean = df_smq_filtered[['SEQN', 'current_smoker']]

print("\n" + "="*50)
print("Excluded former smokers:", len(df_smq) - len(df_smq_filtered))
print("="*50 + "\n")

# 4. MEAN ATTACHMENT LOSS: mask invalid site codes, require >=14 valid sites
loa_cols = [col for col in df_ohx.columns if 'OHX' in str(col).upper() and 'LA' in str(col).upper()]

df_ohx_clean_sites = df_ohx[loa_cols].copy()
for col in loa_cols:
    df_ohx_clean_sites[col] = np.where(df_ohx_clean_sites[col] > 12, np.nan, df_ohx_clean_sites[col])

df_ohx['valid_sites'] = df_ohx_clean_sites.count(axis=1)
df_ohx['mean_LOA_mm'] = df_ohx_clean_sites.mean(axis=1)
df_ohx_clean = df_ohx[df_ohx['valid_sites'] >= 14][['SEQN', 'mean_LOA_mm']]

# 5. CONSOLIDATE MATRIX AND RUN REGRESSION MODEL
print("Merging data frames into master cohort...")
df_master = df_demo.merge(df_ghb, on='SEQN') \
                   .merge(df_smq_clean, on='SEQN') \
                   .merge(df_ohx_clean, on='SEQN')

# Drop any remaining missing data to finalize the pure cohort
df_master = df_master.dropna().reset_index(drop=True)

print(f"Executing OLS engine on clean sample (n = {len(df_master)})...")
model_formula = "mean_LOA_mm ~ LBXGH * current_smoker + RIDAGEYR + RIAGENDR"
results = smf.ols(formula=model_formula, data=df_master).fit()

print("\n" + "="*50 + "\nFINAL OLS REGRESSION RESULTS\n" + "="*50)
print(results.summary())
```
